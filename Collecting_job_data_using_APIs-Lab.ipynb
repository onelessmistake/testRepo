```python
def load_data(): #Function to call the api 
    api_url = ("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json####")
    response = requests.get(api_url)
    if response.status_code == 200:
        print(f"Data Retrieved with {response.status_code} code")
        job_data = response.json()
    else:
        print(f"Data Did not retrieved {response.status_code}") #Debugging to see if the data got retrieved
    return pd.DataFrame(job_data) #Store the data into columns in pandas DataFrame
df = load_data() # store the data to a variable called df

def get_number_of_jobs_T(df, technology): #has the df input and technology
    #putting inside not only characters but special characters like + (C++ for example) and checking for unique.
    #for example if we search for C language we get C,CSS,Scala and not only the C
    t_lower = technology.lower()
    if re.match(r"^[A-Za-z0-9]+$",technology):
        t_pattern = rf"\b{re.escape(technology)}\b"
    else:
        t_pattern = re.escape(technology)
    #make a tech variable which stores the search from a specific column in DataFrame. We can make another like
    # t_mask2 = df['Job Title'].str.contains(t_pattern, case = False, na = False, regex = True) and add it in the
    # number_of_jobs_t like this int(t_mask1.sum() + t_mask2.sum())
    t_mask1 = df['Key Skills'].str.contains(t_pattern, case = False, na = False, regex = True)
    
    number_of_jobs_t = int(t_mask1.sum())
    
    return technology, number_of_jobs_t
```

Calling the function for Python and checking if it works.


```python
print(get_number_of_jobs_T(df,"c++"))
```

#### Write a function to find number of jobs in US for a location of your choice


```python
 def get_number_of_jobs_L(df, location):
    #same as the number for tech jobs
    if re.match(r"^[A-Za-z0-9]+$",location):
        l_pattern = rf"\b{re.escape(location)}\b"
    else:
        l_pattern = re.escape(location)
        
    l_mask1 = df['Location'].str.contains(l_pattern, case = False, na = False, regex = True)
    
    number_of_jobs_l = int(l_mask1.sum())
    
    return location,number_of_jobs_l
```
